```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Bidirectional, Dense, Dropout, Masking, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras import mixed_precision
import matplotlib.pyplot as plt
from tensorflow.keras.layers import LayerNormalization

# ==========================================
# 1. KIỂM TRA GPU VÀ BẬT MIXED PRECISION
# ==========================================
print("Kiểm tra phần cứng...")
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    print(f"✅ Đã tìm thấy {len(gpus)} GPU: {gpus[0].name}")
    # Cho phép GPU tự động mở rộng bộ nhớ (tránh lỗi Out Of Memory)
    tf.config.experimental.set_memory_growth(gpus[0], True)
else:
    print("❌ KHÔNG TÌM THẤY GPU! Hãy bật GPU trong phần Settings (Accelerator) của Kaggle.")

# Bật Mixed Precision (Tăng tốc độ x2 trên GPU NVIDIA)
policy = mixed_precision.Policy('mixed_float16')
mixed_precision.set_global_policy(policy)
print(f"🚀 Chế độ tính toán: {policy.compute_dtype} | Lưu trữ: {policy.variable_dtype}")

# ==========================================
# 2. LOAD DATA VÀ TẠO PIPELINE TỐI ƯU TỐC ĐỘ (TF.DATA)
# ==========================================
DATA_DIR = "/kaggle/input/datasets/huuthang8686/dulieutrain/processed_data"

print("\nĐang tải dữ liệu...")
X_train = np.load(f"{DATA_DIR}/X_train.npy")
y_train = np.load(f"{DATA_DIR}/y_train.npy")
X_val = np.load(f"{DATA_DIR}/X_val.npy")
y_val = np.load(f"{DATA_DIR}/y_val.npy")

SEQ_LENGTH = X_train.shape[1] # 60
FEATURES = X_train.shape[2]   # 70
PAD_VALUE = -99.0

# Vì chạy GPU nên ta có thể tăng Batch Size lên để tận dụng nhân CUDA
BATCH_SIZE = 256  # Bạn có thể thử 256 nếu RAM GPU còn dư
AUTOTUNE = tf.data.AUTOTUNE

# Đóng gói data vào Pipeline để load đa luồng
train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train))
train_dataset = train_dataset.shuffle(buffer_size=10000).batch(BATCH_SIZE).prefetch(AUTOTUNE)

val_dataset = tf.data.Dataset.from_tensor_slices((X_val, y_val))
val_dataset = val_dataset.batch(BATCH_SIZE).prefetch(AUTOTUNE)

# ==========================================
# 3. XÂY DỰNG MÔ HÌNH BILSTM (CHỐNG OVERFITTING)
# ==========================================
model = Sequential([

    Masking(mask_value=PAD_VALUE, input_shape=(SEQ_LENGTH, FEATURES)),

    Bidirectional(LSTM(
        64,
        return_sequences=True,
        dropout=0.3,
        recurrent_dropout=0.2,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )),
    LayerNormalization(),
    Dropout(0.3),

    Bidirectional(LSTM(
        32,
        dropout=0.3,
        recurrent_dropout=0.2,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )),
    LayerNormalization(),
    Dropout(0.4),

    Dense(16, activation='relu'),
    Dropout(0.3),

    Dense(1, activation='sigmoid', dtype='float32')

])

# Giảm nhẹ tốc độ học ban đầu để mô hình đi chậm mà chắc
optimizer = tf.keras.optimizers.Adam(learning_rate=0.0005)

model.compile(
    optimizer=optimizer,
    loss='binary_crossentropy',
    metrics=[
        'accuracy', 
        tf.keras.metrics.AUC(name='auc'),
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall')
    ]
)
# ==========================================
# 4. QUẢN LÝ HUẤN LUYỆN (CALLBACKS)
# ==========================================
MODEL_SAVE_PATH = "/kaggle/working/best_bilstm_model.h5"

callbacks = [
    EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True, verbose=1),
    ModelCheckpoint(filepath=MODEL_SAVE_PATH, monitor='val_loss', save_best_only=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=1)
]

# ==========================================
# 5. TRAINING VỚI GPU PIPELINE
# ==========================================
print("\n🔥 BẮT ĐẦU ĐỐT GPU... 🔥")
history = model.fit(
    train_dataset,              # Đưa pipeline vào thay vì X_train, y_train
    validation_data=val_dataset, # Đưa pipeline val vào
    epochs=100, 
    callbacks=callbacks,
    verbose=1
)
# ==========================================
# 6. VẼ LỊCH SỬ HUẤN LUYỆN (ĐÃ FIX LỖI)
# ==========================================
print("\nĐang vẽ biểu đồ...")
# SỬA LỖI: Tạo lưới 3 hàng, 2 cột (chứa được 6 biểu đồ) và tăng chiều cao ảnh lên 15
fig, axes = plt.subplots(3, 2, figsize=(15, 15))

# Loss (Hàng 1, Cột 1)
axes[0, 0].plot(history.history['loss'], label='Train Loss', color='blue')
axes[0, 0].plot(history.history['val_loss'], label='Val Loss', color='red')
axes[0, 0].set_title('Loss (Càng thấp càng tốt)')
axes[0, 0].legend()

# Accuracy (Hàng 1, Cột 2)
axes[0, 1].plot(history.history['accuracy'], label='Train Acc', color='blue')
axes[0, 1].plot(history.history['val_accuracy'], label='Val Acc', color='red')
axes[0, 1].set_title('Accuracy')
axes[0, 1].legend()

# AUC (Hàng 2, Cột 1)
axes[1, 0].plot(history.history['auc'], label='Train AUC', color='blue')
axes[1, 0].plot(history.history['val_auc'], label='Val AUC', color='red')
axes[1, 0].set_title('AUC Score')
axes[1, 0].legend()

# Recall (Hàng 2, Cột 2)
axes[1, 1].plot(history.history['recall'], label='Train Recall', color='blue')
axes[1, 1].plot(history.history['val_recall'], label='Val Recall', color='red')
axes[1, 1].set_title('Recall (Độ nhạy bắt té ngã)')
axes[1, 1].legend()

# Precision (Hàng 3, Cột 1) - BÂY GIỜ ĐÃ HỢP LỆ!
axes[2, 0].plot(history.history['precision'], label='Train Precision', color='blue')
axes[2, 0].plot(history.history['val_precision'], label='Val Precision', color='red')
axes[2, 0].set_title('Precision')
axes[2, 0].legend()

# Ẩn ô biểu đồ trống ở (Hàng 3, Cột 2) cho đẹp mắt
fig.delaxes(axes[2, 1])

plt.tight_layout()
plt.savefig('/kaggle/working/training_history.png', dpi=300)
print("✅ Hoàn tất! Đã lưu best_model.keras và ảnh training_history.png")
```
